```python
from sklearn.cluster import HDBSCAN
import numpy as np
import pandas as pd
from scipy.spatial.distance import jensenshannon, cosine, euclidean, cityblock

cleanp = "data/clean/"
filname = "exemplo1.csv"
procp = "data/analysed/"
rawp = "data/raw/"

cl_cos = HDBSCAN(store_centers="both",min_cluster_size=4, copy=True, metric=cosine)
cl_eu = HDBSCAN(store_centers="both",min_cluster_size=4, copy=True, metric=euclidean)
cl_ct = HDBSCAN(store_centers="both",min_cluster_size=4, copy=True, metric=cityblock)
cl_js = HDBSCAN(store_centers="both",min_cluster_size=4, copy=True, metric=jensenshannon)


std_data = pd.read_csv(cleanp+"std_"+filname, index_col=0)
rsc_data = pd.read_csv(cleanp+"rsc_"+filname, index_col=0)
prob_data = pd.read_csv(cleanp+"prob_"+filname, index_col=0)
clean_data = pd.read_csv(cleanp+filname, index_col=0)

cl_cos.fit(std_data)
clean_data["StScaler-Cosine labels"] = cl_cos.labels_ #separa bem, mas pega bastante noise
cl_cos.fit(rsc_data)
clean_data["RobScaler-Cosine labels"] = cl_cos.labels_ #ruim, melhor qdo há noise??
cl_eu.fit(rsc_data)
clean_data["RobScaler-Euclidean labels"] = cl_eu.labels_ #bom qdo pouco noise, piora com 0.2
cl_eu.fit(std_data)
clean_data["StScaler-Euclidean labels"] = cl_eu.labels_ #ruim
cl_ct.fit(rsc_data)
clean_data["RobScaler-manhattan labels"] = cl_ct.labels_ #bom qdo pouco noise
cl_ct.fit(std_data)
clean_data["StScaler-manhattan labels"] = cl_ct.labels_ #ruim
cl_js.fit(prob_data)
clean_data["ProbScaler labels"] = cl_js.labels_ #ótimo

clean_data.to_csv(procp+filname, index=True)
```

```python
##
lista = []
logpath = f"data/log_{filname}.txt"
log = open(logpath, "a", encoding="utf-8")
for i in range(100, 107): lista.append(clean_data.iloc[:,i].value_counts())
log.write(f"{lista}\n")
```
