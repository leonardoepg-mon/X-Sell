```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, RobustScaler

rawp = "data/raw/"
cleanp = "data/clean/"
filname = "exemplo1"
logpath = f"data/clean/log_{filname}.txt"
log = open(logpath, "w", encoding="utf-8")
df = pd.read_csv(rawp+filname+".csv", index_col=0)
df.fillna(0)
log.write(f"Total de linhas inválidas: {df.sum(1).size - np.count_nonzero(df.sum(1))} \n")
df = df.loc[df.sum(1) != 0]
log.write(f"Total de colunas inválidas: {df.sum(0).size - np.count_nonzero(df.sum(0))} \n")
df = df.loc[:,df.sum(0) != 0]
df.to_csv(cleanp+filname+".csv", index=True)
sscaler = StandardScaler(copy=True)
std_cl = pd.DataFrame(sscaler.fit_transform(df))
std_cl.to_csv(cleanp+"std_"+filname+".csv", index = True)

rscaler = RobustScaler(copy=True)
rob_cl = pd.DataFrame(rscaler.fit_transform(df))
rob_cl.to_csv(cleanp+"rsc_"+filname+".csv", index=True)

prob_cl = df.div(df.sum(1), axis=0)
prob_cl.to_csv(cleanp+"prob_"+filname+".csv", index=True)
```
